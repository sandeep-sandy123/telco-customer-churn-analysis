```python
import pandas as pd
df = pd.read_csv(r"C:\Users\user\Downloads\Telco-Customer-Churn.csv")
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')
print(df['TotalCharges'].isnull().sum())
```

```python
print(df[df['TotalCharges'].isnull()][['tenure', 'MonthlyCharges', 'TotalCharges', 'Churn']])
```

```python
df['TotalCharges'] = df['TotalCharges'].fillna(0)

# Verify it's clean
print("Nulls remaining:", df['TotalCharges'].isnull().sum())
print("Total rows:", len(df))
```

```python
churn_by_contract = df.groupby('Contract')['Churn'].apply(
    lambda x: (x == 'Yes').sum() * 100 / len(x)
).round(2)

print(churn_by_contract.sort_values(ascending=False))
```

```python
churn_by_security = df.groupby('OnlineSecurity')['Churn'].apply(
    lambda x: (x == 'Yes').sum() * 100 / len(x)
).round(2)

print(churn_by_security.sort_values(ascending=False))
```

```python
import sqlite3

conn = sqlite3.connect("telco.db")
df.to_sql("telco", conn, if_exists="replace", index=False)
print("Database created successfully")
```

```python
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce').fillna(0)
df.to_csv("Telco_Cleaned.csv", index=False)
print("Done. Rows:", len(df))
print("Nulls in TotalCharges:", df['TotalCharges'].isnull().sum())
```
